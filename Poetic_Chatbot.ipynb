#Writing the persona prompt

```python
POET_PERSONA_PROMPT = """
You are Pri, a gentle and thoughtful poet-companion. You never speak in plain
prose.Every reply you give must be a short original poem responding to what
the user said.

Rules:
- Reply ONLY with the poem. no greetings, no explanation, no "Here is a poem".
- Keep it to 2-4 short lines.
- Match the emotional tone of the message (happy -> bright imagery, sad -> soft
imagery, angry -> stormy but never cruel imagery).

Example:
User: I'm feeling happy today
Pri: In fields of joy, your heart does dance,
With sunlight's glow, your soul's expanse.
"""

```

#Turning the prompt into a Reusable Function

```python
def get_poetic_response(user_message):
    """
    Sends the user's message to Groq, wrapped in the poetic persona prompt,
    and returns just the poem text. Falls back gracefully on error.
    """
    try:
        response = client.chat.completions.create(
            model="openai/gpt-oss-20b",
            messages=[
                {"role": "system", "content": POET_PERSONA_PROMPT},
                {"role": "user", "content": user_message}
            ]
        )
        poem = response.choices[0].message.content.strip()
        return poem
    except Exception as e:
        print(f"[DEBUG] Error was: {e}")   # remove this line before final submission
        return "The muse grows quiet, spent and still,\nGive me a moment, then speak your will."
```

```python
conversation_history = []

def get_poetic_response(user_message):
    conversation_history.append({"role": "user", "content": user_message})

    messages = [{"role": "system", "content": POET_PERSONA_PROMPT}] + conversation_history

    try:
        response = client.chat.completions.create(
            model="openai/gpt-oss-20b",
            messages=messages
        )
        poem = response.choices[0].message.content.strip()
        conversation_history.append({"role": "assistant", "content": poem})
        return poem
    except Exception as e:
        return "The muse grows quiet, spent and still,\nGive me a moment, then speak your will."
```
